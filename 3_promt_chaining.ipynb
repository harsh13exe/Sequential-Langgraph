```python
from langgraph.graph import StateGraph, START,END
from langchain_ollama import ChatOllama
from typing import TypedDict
```

```python
model = ChatOllama(model = "llama3.2:3b",temperature=0.3)
```

```python
class BlogState(TypedDict):
    title:str
    outline:str
    content:str
    rate:int
```

```python
def create_outline(state:BlogState)->BlogState:
    #fetch title
    title = state['title']
    
    #call llm gen outline
    promt = f"Generate an outline on the topic - {title}"
    outline = model.invoke(promt).content
    #update state
    state['outline'] = outline
    return state
```

```python
def create_blog(state:BlogState)->BlogState:
    #fetch title and outline
    title = state['title']
    outline = state['outline']
    
    #call llm gen outline
    promt = f"Generate an outline on the topic - {title} using following outline - {outline}"
    content = model.invoke(promt).content
    #update state
    state['content'] = content
    return state
```

```python
def rate_blog(state:BlogState)->BlogState:
    #fetch title and outline
    content = state['content']
    
    promt = f"Rate this blog - {content}"
    rate = model.invoke(promt).content
    #update state
    state['rate'] = rate
    return state
```

```python
graph = StateGraph(BlogState)

graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
graph.add_node('rate_blog',rate_blog)

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','rate_blog')
graph.add_edge('rate_blog',END)

workflow = graph.compile()
```

```python
inital_state = {'title':'Raise of AI in india'}
final_state = workflow.invoke(inital_state)
print(final_state)
```
